### Hướng dẫn 3 bước:

```python
# [BƯỚC 2] Chuẩn bị mã nguồn dự án
import os
from google.colab import files

if os.path.exists("aov_checker.zip"):
    !unzip -o aov_checker.zip
    print("✅ Đã giải nén aov_checker.zip thành công!")
elif os.path.exists("main.py") and os.path.exists("buildozer.spec"):
    print("✅ Mã nguồn đã sẵn sàng trong thư mục hiện tại!")
else:
    print("📂 Vui lòng chọn file aov_checker.zip từ máy tính của bạn để tải lên:")
    uploaded = files.upload()
    for fname in uploaded.keys():
        if fname.endswith(".zip"):
            !unzip -o "{fname}"
            print(f"✅ Đã giải nén {fname} thành công!")
```

```python
# [BƯỚC 3] Tiến hành biên dịch APK (Buildozer Debug)
# Lần đầu chạy sẽ mất khoảng 10-15 phút để Google Colab tự tải Android SDK & NDK
!yes | buildozer -v android debug
```

```python
# [BƯỚC 4] Tìm và tự động tải file APK về máy tính/điện thoại
import glob
from google.colab import files

apk_files = glob.glob("bin/*.apk")
if apk_files:
    apk_path = apk_files[0]
    print(f"🎉 CHÚC MỪNG! Đã tạo thành công APK: {apk_path}")
    print("Đang tải file APK xuống máy tính của bạn...")
    files.download(apk_path)
else:
    print("❌ Không tìm thấy file APK trong thư mục bin/. Vui lòng kiểm tra lại log lỗi ở Bước 3!")
```
